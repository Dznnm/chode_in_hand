```python
def solve_quad_eqn(a,b,c):
  root1= ((-b+(b**2 - 4*a*c)**0.5)/2*a)
  root2= ((-b-(b**2 - 4*a*c)**0.5)/2*a)
  roots= (root1,root2)
  return roots
print(solve_quad_eqn(1,-5,6))
```

```python
def sum_oddrange(odd):
  nem=0
  for nom in range(odd+1):
    if nom % 2 != 0:
      nem = nem + nom
  return nem
```

```python
print(sum_oddrange(100))
```

```python
def sum_evenrange(even):
  nem=0
  for nom in range(even+1):
    if nom % 2 == 0:
      nem = nem + nom
  return nem
```

```python
print(sum_evenrange(100))
```
