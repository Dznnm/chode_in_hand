```python
numbers=(0,1,2,3,4,5)
for number in numbers:
  print(number)
  if number == 3:
    continue
  print('Next number should be',number + 1) if number != 5 else print("loop's end")
print('outside the loop')
```

```python
#Range Function
lst=list(range(11))
print(lst)
st=set(range(1,11))
print(st)

lst=list(range(0,11,2))
print(lst)
st=set(range(0,11,2))
print(st)

#for backward from start to end
lst=list(range(11,0,-2))
print(lst)
```
