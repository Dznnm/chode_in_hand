### Manual Counting and Sorting for Top 10 Languages

```python
all_languages_manual = []
for country_info in countries_data:
  languages = country_info.get('languages', [])
  all_languages_manual.extend(languages)

# 1. Manual Counting using a dictionary
language_counts_manual = {}
for lang in all_languages_manual:
  language_counts_manual[lang] = language_counts_manual.get(lang, 0) + 1

# 2. Sorting the dictionary items by count in descending order
# We convert the dictionary items to a list of (language, count) tuples
sorted_languages = sorted(language_counts_manual.items(), key=lambda item: item[1], reverse=True)

# 3. Taking the top 10
top_10_languages_manual = sorted_languages[:10]

print("The ten most spoken languages (manual method) are:")
for language, count in top_10_languages_manual:
  print(f"- {language}: {count} countries")
```
