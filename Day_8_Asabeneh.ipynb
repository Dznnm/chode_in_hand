```python
#Dictionariesss
#Unordered, modifiable, paired (key:value)

empty_dict={}
about_me={
    'first_name':'Dzannun',
    'last_name':'Muhlashon',
    'age':'250',
    'country':'Indonesia',
    'is_married':False,
    'skills':['fitness','analytical thinking','excel','AI'],
    'address':{
        'street':'pucangsimo street',
        'zipcode':'61462'
    }
}
```

```python
len(about_me)
print(about_me)
print(about_me['address']['zipcode'])
print(about_me['age'])
print(about_me['skills'][2])
print(about_me.get('is_married'))
print(about_me.get('city'))
```

```python
about_me['height']='162 cm'
```

```python
print(about_me['height'])
about_me['job_title']='QC staff'
about_me['skills'].append('inspection')
print(about_me['skills'])
print(about_me['job_title'])
```

```python
#modifying dict
about_me['age']='25 years old'
print(about_me['age'])
about_me['skills'].remove('inspection')
```

```python
print(about_me['skills'])
#removing key and value pairs from a dict
about_me.pop('first_name')
about_me.popitem()
del about_me['is_married']
```
